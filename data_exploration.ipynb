# FLEMMS 2024 Data Exploration Notebook

This notebook provides comprehensive exploration of the FLEMMS (Family Living and Expenditure Microsimulation Model System) 2024 datasets.

## Dataset Overview
- **Volume 1**: Contains household, member, and RTF data for Volume 1
- **Volume 2**: Contains household, member, and RTF data for Volume 2
- **File Types**: CSV files with corresponding metadata dictionaries
- **Data Structure**: Household-level and individual-level data with expenditure information

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path
import warnings

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
warnings.filterwarnings('ignore')

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

print("Libraries imported successfully!")
print(f"Working directory: {os.getcwd()}")
```

## 1. Data Loading and Initial Exploration

```python
# Define data paths
data_dir = Path("data")
vol1_dir = data_dir / "PHL-PSA-FLEMMS-2024-V1-PUF"
vol2_dir = data_dir / "PHL-PSA-FLEMMS-2024-V2-PUF"

print("Data directories:")
print(f"Volume 1: {vol1_dir}")
print(f"Volume 2: {vol2_dir}")
print(f"Volume 1 exists: {vol1_dir.exists()}")
print(f"Volume 2 exists: {vol2_dir.exists()}")
```

```python
# Load Volume 1 datasets
print("Loading Volume 1 datasets...")

vol1_household = pd.read_csv(vol1_dir / "FLEMMS PUF 2024 Volume1 - HOUSEHOLD.CSV")
vol1_member = pd.read_csv(vol1_dir / "FLEMMS PUF 2024 Volume1 - MEMBER.CSV")
vol1_rtf1 = pd.read_csv(vol1_dir / "FLEMMS PUF 2024 Volume1 - RTF1.CSV")
vol1_rtf2 = pd.read_csv(vol1_dir / "FLEMMS PUF 2024 Volume1 - RTF2.CSV")

print(f"Volume 1 Household data shape: {vol1_household.shape}")
print(f"Volume 1 Member data shape: {vol1_member.shape}")
print(f"Volume 1 RTF1 data shape: {vol1_rtf1.shape}")
print(f"Volume 1 RTF2 data shape: {vol1_rtf2.shape}")
```

```python
# Load Volume 2 datasets
print("Loading Volume 2 datasets...")

vol2_household = pd.read_csv(vol2_dir / "FLEMMS PUF 2024 Volume2 - HOUSEHOLD.CSV")
vol2_member = pd.read_csv(vol2_dir / "FLEMMS PUF 2024 Volume2 - MEMBER.CSV")
vol2_rtf1 = pd.read_csv(vol2_dir / "FLEMMS PUF 2024 Volume2 - RTF1.CSV")
vol2_rtf3 = pd.read_csv(vol2_dir / "FLEMMS PUF 2024 Volume2 - RTF3.CSV")

print(f"Volume 2 Household data shape: {vol2_household.shape}")
print(f"Volume 2 Member data shape: {vol2_member.shape}")
print(f"Volume 2 RTF1 data shape: {vol2_rtf1.shape}")
print(f"Volume 2 RTF3 data shape: {vol2_rtf3.shape}")
```

## 2. Basic Data Structure Analysis

```python
# Display column names for each dataset
datasets = {
    'Vol1_Household': vol1_household,
    'Vol1_Member': vol1_member,
    'Vol1_RTF1': vol1_rtf1,
    'Vol1_RTF2': vol1_rtf2,
    'Vol2_Household': vol2_household,
    'Vol2_Member': vol2_member,
    'Vol2_RTF1': vol2_rtf1,
    'Vol2_RTF3': vol2_rtf3
}

for name, df in datasets.items():
    print(f"\n{name} columns ({len(df.columns)} columns):")
    print(df.columns.tolist())
```

```python
# 1. Analyze Data Types and Missing Values for Proposal Features
print("--- PROPOSAL FEATURES SUMMARY ---\n")

# Mapping the datasets to the specific columns you care about
target_features = {
    'Vol1_Household': ['REG', 'REG2', 'PRV', 'HHID', 'URBANITY'],
    'Vol1_Member': ['LNO', 'SEX', 'AGE', 'HGC_LEVEL', 'ATTEND_SCHOOL', 'NATT_REASON', 'ETHNICITY'],
    'Vol1_RTF1': ['ELECTRICITY'],
    'Vol1_RTF2': ['FLITERATE', 'RESP_RFACT_F2', 'AGE_F2', 'SEX_F2', 'HGC_GRADE_F2'],
    'Vol2_RTF3': ['INTERNET']
}

for name, cols in target_features.items():
    df = datasets[name]
    # Only check columns that actually exist in the dataframe
    valid_cols = [c for c in cols if c in df.columns]
    
    if valid_cols:
        summary = pd.DataFrame({
            'DType': df[valid_cols].dtypes,
            'Missing_Count': df[valid_cols].isnull().sum(),
            '%_Missing': (df[valid_cols].isnull().sum() / len(df) * 100).round(2),
            'Unique_Values': df[valid_cols].nunique()
        })
        print(f"[{name}]")
        print(summary)
        print("-" * 50)

# 2. Investigate your "Missing/Question" Variables
print("\n--- INVESTIGATING UNKNOWN VARIABLES ---")

# Check if ETHNICITY can be used as a proxy for "Language Spoken at Home"
if 'ETHNICITY' in vol1_member.columns:
    print("\nTop 5 values for ETHNICITY (Possible proxy for Language):")
    print(vol1_member['ETHNICITY'].value_counts().head())

# Look for HUC / PSGC clues in the Household dataset
print("\nGeographic indicators available in Household Data:")
geo_cols = [col for col in vol1_household.columns if col in ['REG', 'REG2', 'PRV', 'MUN', 'BGY', 'PSU', 'URBANITY']]
print(geo_cols)
if 'PRV' in vol1_household.columns:
    print(f"\nNumber of unique Provinces/HUCs (PRV): {vol1_household['PRV'].nunique()}")
```

```python
import pandas as pd

# 1. Track where every unique column appears across all datasets
column_locations = {}
total_datasets = len(datasets)

for df_name, df in datasets.items():
    for col in df.columns:
        if col not in column_locations:
            # Save the datatype and start a list of where we found it
            column_locations[col] = {'datasets': [], 'dtype': str(df[col].dtype)}
        column_locations[col]['datasets'].append(df_name)

# 2. Format the Location string to match your proposal table style
formatted_data = []

for col, info in column_locations.items():
    present_in = info['datasets']
    
    # Logic to aggregate and format the Location string
    if len(present_in) == total_datasets and total_datasets > 0:
        loc_str = "Present in all"
    else:
        # Join the names nicely (e.g., "Vol1 - Household, Vol2 - Household")
        loc_str = ", ".join(present_in).replace("_", " - ")
        
    formatted_data.append({
        'FEATURE NAME': col,
        'Label/Description': '', # Blank for now so you can fill it in later if needed
        'DType': info['dtype'],
        'Location': loc_str
    })

# 3. Create the final DataFrame
master_summary_table = pd.DataFrame(formatted_data)

# 4. Save to a CSV file
csv_filename = 'FLEMMS_Aggregated_Master_Dictionary.csv'
master_summary_table.to_csv(csv_filename, index=False)

# Print the results
print(f"Total unique variables found across all datasets: {len(master_summary_table)}")
print(f"File successfully saved as: {csv_filename}\n")

# Preview the first 20 rows
print(master_summary_table.head(20).to_markdown(index=False))
```
